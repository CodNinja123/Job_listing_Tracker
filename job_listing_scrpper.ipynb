```python
import requests
from bs4 import BeautifulSoup
import csv

URL = "https://realpython.github.io/fake-jobs/"
CSV_FILE = "python_job_listings.csv"

def scrape_and_save_jobs():
    print(f"Connecting to {URL}...")
    response = requests.get(URL)
    
    if response.status_code != 200:
        print(f"Error: Could not retrieve webpage. Status code: {response.status_code}")
        return

    soup = BeautifulSoup(response.text, "html.parser")
    
    # Robust Check 1: Ensure the container exists before processing
    results_container = soup.find(id="ResultsContainer")
    if not results_container:
        print("Error: 'ResultsContainer' element was not found in the HTML.")
        return

    # Extract all individual job cards
    job_elements = results_container.find_all("div", class_="card-content")
    print(f"Successfully discovered {len(job_elements)} job postings.")

    # Open a new CSV file to write our findings
    with open(CSV_FILE, mode="w", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)
        
        # Write the header row
        writer.writerow(["Job Title", "Company Name", "Location", "Job Detail Page URL"])
        
        saved_count = 0
        for job in job_elements:
            # Robust Check 2: Safely extract each field and handle missing elements gracefully
            title_tag = job.find("h2", class_="title")
            company_tag = job.find("h3", class_="company")
            location_tag = job.find("p", class_="location")
            
            # The page has two links; the second link ("Apply") contains the unique detail URL
            links = job.find_all("a")
            
            # Populate text fields only if the tags exist, otherwise use fallback text
            title = title_tag.text.strip() if title_tag else "N/A"
            company = company_tag.text.strip() if company_tag else "N/A"
            location = location_tag.text.strip() if location_tag else "N/A"
            url = links[1]["href"].strip() if len(links) >= 2 else "N/A"
            
            # Write the clean data row to our CSV file
            writer.writerow([title, company, location, url])
            saved_count += 1

    print(f"🚀 Success! Cleaned data for {saved_count} jobs saved to '{CSV_FILE}'")

# Run the complete scraping pipeline
scrape_and_save_jobs()
```
